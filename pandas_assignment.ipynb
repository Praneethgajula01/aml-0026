```python
"""
Create the following DataFrame:

| Name  | Age | Department | Marks |
| ----- | --: | ---------- | ----: |
| Asha  |  22 | CSE        |    85 |
| Ravi  |  23 | ECE        |    78 |
| Meena |  21 | CSE        |    92 |
| Kiran |  24 | ME         |    74 |
| Arun  |  22 | ECE        |    88 |

Display:

	Complete DataFrame
	First 3 rows
	Last 2 rows
	Shape
	Column names
	Data types
"""


df=pd.DataFrame({'Name':['Asha', 'Ravi', 'Meena', 'Kiran', 'Arun'],
                 'Age':[22, 23, 21, 24, 22],
                 'Department':['CSE', 'ECE', 'CSE', 'ME', 'ECE'],
                 'Marks':[85, 78, 92, 74, 88]})
print(df)

print(df[:3])

print(df[-2:])

print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.info())
print(df.describe())
```

```python
"""
Using the previous DataFrame, display:

	info()
	describe()
	Number of rows
	Number of columns
"""



print(df.info())
print(df.describe())
print(df.shape[0])
print(df.shape[1])
```

```python
"""
Find:

	Students scoring above 80.
	Students scoring between 75 and 90.
	Students belonging to CSE.
	CSE students scoring above 80.
"""

print(df[df["Marks"]>80])
print(df[(df["Marks"]>75) & (df["Marks"]<90)])


cse_df=df[df["Department"]=="CSE"]
print(cse_df)
print(cse_df[cse_df["Marks"]>80])

```

```python
#Display students belonging to either CSE or ECE.

print(df[(df["Department"]=="CSE") | (df["Department"]=="ECE")])
```

```python
#9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".

df["Result"]=["Pass" if marks>=75 else "Fail" for marks in df["Marks"]]

df
```

```python
"""
10. Assign grades according to:

	| Marks | Grade |
	| ----: | ----- |
	|  ≥ 90 | A     |
	|  ≥ 80 | B     |
	|  ≥ 70 | C     |
	|  < 70 | D     |
"""

df["Grade"] = ["A" if marks >= 90 else "B" if marks >= 80 else "C" if marks >= 70 else "D" for marks in df["Marks"]]

df
```

```python
"""
Sort students:

	By marks in ascending order.
	By marks in descending order.
	By Department and then Marks.
"""

print(df.sort_values(by="Marks"))
print(df.sort_values(by="Marks", ascending=False))
print(df.sort_values(by=["Department", "Marks"]))
```

```python
"""
Consider:

data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

"""
import numpy as np

data= {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students=pd.DataFrame(data)

print(students.notna())

print("no of missing values in each value:")

for col in data:
    print(f"{col}: {students[col].isna().sum()}")

print("rows containing missing values:")
print(students.isna().sum(axis=1))
print(students.isna().any(axis=1))
```

```python
"""
13. Replace:

	Missing Age with average Age
	Missing Marks with average Marks
"""

students["Age"].fillna(students["Age"].mean())
students["Marks"].fillna(students["Marks"].mean())
print(students)
```

```python
"""
14. Remove rows containing missing values.

"""

print(students.dropna())
```

```python
"""
16. Given:

marks = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 92, 67, 88]
})

Calculate:

Mean
Median
Maximum
Minimum
Standard deviation
Sum
"""

marks=pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 92, 67, 88]
})

print("Mean:", marks["Marks"].mean())
print("Median:", marks["Marks"].median())
print("Max:", marks["Marks"].max())
print("Min:", marks["Marks"].min())
print("std:", int(marks["Marks"].std()))
print("sum:", marks["Marks"].sum())
```

```python
"""
17. Consider:

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)


Find:
Average marks by department
Maximum marks by department
Number of students in each department

For each department, calculate:
Mean
Minimum
Maximum
Count
"""

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df=pd.DataFrame(data)
print(df.groupby("Department")["Marks"].mean())
print(df.groupby("Department")["Marks"].max())
print(df.groupby("Department")["Marks"].count())

#by using agg function

print(df.groupby("Department")["Marks"].agg(["mean", "min", "max", "count"]))
```

```python
import numpy as np

data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)

df

#first five

print(df[0:5])
print(df.ndim)

#Find the average final mark.
print(df.FinalMarks.mean().astype(int))

#Find the highest-scoring student.
print(df.FinalMarks.max())

#Find the highest-scoring student.
print(df.FinalMarks.min())

#students scoring above 80
print(df[df.FinalMarks>80])

#Find average marks by department.
print(df.groupby("Department")["FinalMarks"].mean())

#students with attendance above 80%
print(df[df["Attendance"]>80])

#sort students by final marks
print(df.sort_values(by="FinalMarks"))

#result column for pass or fail
df["Result"]=["pass" if marks>=80 else "fail" for marks in df["FinalMarks"]]
df

# performance category

conditions=[ df["FinalMarks"]>=90, df["FinalMarks"]>=80, df["FinalMarks"]>=70]
category=["Excellent", "Very Good", "Good"]


df["Performance"]=np.select(conditions, category, default="Needs improvement")

df


#dept having highest avg final marks
print(df.groupby("Department")["FinalMarks"].mean())
print(df.groupby("Department")["FinalMarks"].mean().idxmax())
```

```python
"""
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})
"""

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

print(employees["Salary"].mean())

#max and min salary
print(employees["Salary"].min())
print(employees["Salary"].max())


#Employee receiving the highest salary.
print(employees[employees["Salary"]==employees["Salary"].max()])

#above 60k

print(employees[employees["Salary"]>60000])

#avg salary for each dept
print(employees.groupby("Department")["Salary"].mean().round(2))

#avg exp
print(employees.groupby("Department")["Experience"].mean().round(2))

#no of emp in each dept
print(employees.groupby("Department")["Employee"].count())

#sort employees by salary

print(employees.sort_values(by="Salary", ascending=False))


#Find IT employees with more than 3 years of experience.
print(employees[(employees["Department"]=="IT") & (employees["Experience"]>3)])
```
