```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from google.colab import drive
drive.mount('/content/drive')
```

```python
import os
os.listdir('/content/drive/MyDrive')
```

```python
df=pd.read_csv('/content/drive/MyDrive/income_data.csv')
display(df.head())
```

```python
df.columns.values[0]="id"
df
```

```python
X=df[[ "income"]]
y=df["happiness"]
df.isna().sum()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test=train_test_split(
    X, y, test_size=0.3, random_state=42
)

```

```python
from sklearn.linear_model import LinearRegression

model=LinearRegression()
model.fit(X_train, y_train)
y_pred=model.predict(X_test)
```

```python
print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
print("Actual val:", y_test.values[:10])
print("Pred val:", y_pred[:10])
```

```python
plt.scatter(X_train["income"], y_train, color="red", label="Actual Data")
plt.plot(X_test["income"], y_pred, color="blue", label="Pred data")

plt.xlabel("Income")
plt.ylabel("Happiness")
plt.title("Income vs Happiness")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae=mean_absolute_error(y_test, y_pred)
mse=mean_squared_error(y_test, y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test, y_pred)

print("mae:", mae)

print("mse:", mse)

print("rmse:", rmse)

print("r2:", r2)
```

```python
salary=pd.read_csv('/content/drive/MyDrive/Salary_Data.csv')
display(salary.head())
```

```python
salary.isnull().sum()
```

```python
X=salary["YearsExperience"]
y=salary["Salary"]
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test=train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression

model=LinearRegression()

model.fit(X_train.values.reshape(-1, 1), y_train)
y_pred=model.predict(X_test.values.reshape(-1, 1))

print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
print("Actual Income:", y_test.values)
print("Pred values:", y_pred)
```

```python
mae=mean_absolute_error(y_test, y_pred)
mse=mean_squared_error(y_test, y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test, y_pred)

print("mae:", mae)

print("mse:", mse)

print("rmse:", rmse)

print("r2:", r2)
```

```python
plt.scatter(X_train, y_train, color="purple", label="Actual Data")
plt.scatter(X_test, y_test, color="red")
plt.plot(X_test, y_pred, color="orange", label="Pred data")

plt.xlabel("YearsExp")
plt.ylabel("Salary")
plt.title("yoe vs salary")
plt.legend()
plt.show()
```

```python
experience=np.array([[7.5]])
predicted_salary=model.predict(experience)
print(f"Predicted salary for new experience: {predicted_salary[0]:.2f}")
```

### Testing with New Data

Let's test the trained model with some new 'YearsExperience' data to see its predictions.

```python
new_years_experience = pd.Series([1.8, 5.0, 12.0])

new_years_experience_reshaped = new_years_experience.values.reshape(-1, 1)

new_salary_predictions = model.predict(new_years_experience_reshaped)

new_data_df = pd.DataFrame({
    'YearsExperience': new_years_experience,
    'PredictedSalary': new_salary_predictions
})
display(new_data_df)


plt.scatter(X_train, y_train, color="purple", label="Actual Data")
plt.plot(new_years_experience, new_salary_predictions, color="orange", label="Pred data")
plt.xlabel("new_years_exp")
plt.ylabel("new_pred_sal")
plt.title("new prediction on model")
plt.show()
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeCV

model=make_pipeline(
    StandardScaler(),
    RidgeCV(alphas=[0.01, 0.1, 1, 10, 100])
)
model.fit(X_train.values.reshape(-1, 1), y_train)
ridge_model=model.named_steps["ridgecv"].alpha_
print("Best alpha selected:", ridge_model)
```

```python
x_1=pd.DataFrame([[5]])
y_1=model.predict(x_1)
print(y_1)
```
