**LINEAR** **REGRESSION**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

```

```python
income = pd.read_csv('/content/drive/MyDrive/income_data.csv')
income
```

```python
display(income.head())
```

```python
X=income['happiness'].values.reshape(-1,1)
y=income['income']

print(X)
print(y)
```

```python
X_train, X_test, y_train, y_test=train_test_split(
    X, y, train_size=0.7, random_state=42
)

model=LinearRegression()
model.fit(X_train, y_train)
y_pred=model.predict(X_test)
```

```python
print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
print("Actual val:", y_test.values[:10])
print("Pred val:", y_pred[:10])
```

```python
plt.scatter(X_train, y_train, color="violet", label="Actual Data")
plt.plot(X_test, y_pred, color="red", label="Pred data")

plt.xlabel("Happiness")
plt.ylabel("Income")
plt.title("Happiness vs Income")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

mae=mean_absolute_error(y_test, y_pred)
mse=mean_squared_error(y_test, y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test, y_pred)

print("mae:", mae)

print("mse:", mse)

print("rmse:", rmse)

print("r2:", r2)
```

```python
new_data=np.array([[15]])
predicted_happiness=model.predict(new_data)
print(f"Predicted happiness for new income: {predicted_happiness[0]:.2f}")
```

# **RIDGE REGRESSION**

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeCV, Lasso

ridge_pipeline = make_pipeline(
    StandardScaler(),
    RidgeCV(alphas=[0.01, 0.1, 1, 10, 100])
)
ridge_pipeline.fit(X_train, y_train)
ridge_alpha = ridge_pipeline.named_steps["ridgecv"].alpha_
print("Best alpha selected:", ridge_alpha)
```

# **LASSO** **REGRESSION**

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LassoCV

lasso_pipeline = make_pipeline(
    StandardScaler(),
    LassoCV(alphas=[0.01, 0.1, 1, 10, 100], cv=5)
)

lasso_pipeline.fit(X_train, y_train)

lasso_alpha = lasso_pipeline.named_steps["lassocv"].alpha_

print("Best alpha selected:", lasso_alpha)
```

```python
ridge_pred = ridge_pipeline.predict(X_test)
lasso_pred = lasso_pipeline.predict(X_test)
```

```python
print("Linear regression predictions: ", y_pred[:5])
print("ridge predictions: ", ridge_pred[:5])
print("lasso predictions: ",lasso_pred[:5])
```
