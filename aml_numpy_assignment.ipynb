```python
#Q8

salary = np.array([25000, 32000, 28000, 45000,
                   50000, 38000, 42000, 30000])

print(salary)

print(np.sum(salary))

n=len(salary)
avg=np.sum(salary)/n

print("avg is:", avg)

print(np.max(salary))
print(np.min(salary))

#median - np.median()

print(np.median(salary))

#standard deviation - np.std()

print(np.std(salary ))
```

```python
import numpy as np
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

s_marks=np.sum(marks, axis=1)
print(s_marks)
avg_score=np.mean(marks, axis=1)
print("Average marks of each student", avg_score)
avg_score=np.mean(marks, axis=0)
print("Average marks in each subject", avg_score)
print("student with highest marks:", np.argmax(s_marks))
print("highest mark in each subject:", np.max(marks, axis=0))
```

```python
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])


print("the mean of temp is:", temperature.mean())
print("the max temp is:", np.max(temperature), "the min temp is:", np.min(temperature))

print("the no of days above avg:", len(temperature[temperature>temperature.mean()]))

print("the no of days >34:", len(temperature[temperature>34]))
print("the std:", temperature.std())
```

```python
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("total annual sales:", np.sum(sales))

print("avg:", np.average(sales))

print("highest sales month:", np.argmax(sales)+1)
print("least sales month:", np.argmin(sales)+1)

print(">20k:", len(sales[sales>20000]))

print("% increase from jan to dec:", ((sales[-1]-sales[0])/sales[0])*100)

```
